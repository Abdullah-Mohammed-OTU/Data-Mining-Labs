```python
col_names = ['Number of Frequent Flier miles', '% of time playing video games',
             'L of ice cream consumed weekly', 'Types of people']
# load dataset
dataset = pd.read_csv("/content/drive/MyDrive/Data_Mining_datasets/Lab5Datasets/datingData_training.txt",
                      header=None, names=col_names, sep="\t")

dataset.head()
```

```python
feature_cols = ['Number of Frequent Flier miles', '% of time playing video games',
             'L of ice cream consumed weekly']
target_col = 'Types of people'

# Split dataset into features and target variable

X_train = dataset[feature_cols] # Features
y_train = dataset[target_col] # Target variable
```

```python
# Load the test dataset
col_names = ['Number of Frequent Flier miles', '% of time playing video games',
             'L of ice cream consumed weekly', 'Types of people']
# load dataset
test_dataset = pd.read_csv("/content/drive/MyDrive/Data_Mining_datasets/Lab5Datasets/datingData_test.txt",
                      header=None, names=col_names, sep="\t")

# Split dataset into features and target variable

X_test = test_dataset[feature_cols] # Features
y_test = test_dataset[target_col] # Target variable
```

## Model 1: Decision Tree

```python
# Load libraries
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
# Create Decision Tree classifer object
clf = DecisionTreeClassifier()

# Train Decision Tree Classifer
clf = clf.fit(X_train,y_train)

#Predict the response for test dataset
y_pred = clf.predict(X_test)
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nConfusion matrix:\n", confusion_matrix(y_test, y_pred, labels=["didntLike", "smallDoses", "largeDoses"]))
print("\nClassification report:\n", classification_report(y_test, y_pred))
```

## Model 2: KNN

```python
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
```

```python
# Scaling the data
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# Implementing the KNN
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)
```

```python
y_pred = knn.predict(X_test)
```

```python
#Printing metrics
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nConfusion matrix:\n", confusion_matrix(y_test, y_pred, labels=["didntLike", "smallDoses", "largeDoses"]))
print("\nClassification report:\n", classification_report(y_test, y_pred))
```

## Model 3: Naive-Bayes

```python
# Importing libraries
from sklearn.naive_bayes import GaussianNB
from sklearn.pipeline import Pipeline
```

```python
# Build model pipeline
model = Pipeline([
    ("scaler", StandardScaler()),
    ("nb", GaussianNB())
])
```

```python
# Train
model.fit(X_train, y_train)
```

```python
# Predict
y_pred = model.predict(X_test)
```

```python
#Evaluate
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nConfusion matrix:\n", confusion_matrix(y_test, y_pred, labels=["didntLike", "smallDoses", "largeDoses"]))
print("\nClassification report:\n", classification_report(y_test, y_pred))
```
